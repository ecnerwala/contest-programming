```python
inp = """
OOBNN
OBNNN
KBYKB
OBONY
GOONO
GYNKG
NGNOO
KNBGN
OOKON
GYKOY
""".strip()
assert inp.count('O') == 14
assert inp.count('K') == 6
assert inp.count('G') == 6
assert inp.count('Y') == 5
assert inp.count('N') == 13
assert inp.count('B') == 6

dat = inp.split('\n')
assert len(dat) == 10
assert all(len(r) == 5 for r in dat)
assert len(set(dat) | set(r[::-1] for r in dat)) == 2 * len(dat)

valids = {'OK', 'GB', 'ON', 'YN'}
valids = valids | {a[::-1] for a in valids}
def valid(a,b):
    return a+b in valids
```

```python
from itertools import *

for d in product(*([a] for a in dat)):
    for r0 in d:
        for cols in permutations((r for r in d if r != r0), 5):
            if not all(valid(r0[i],cols[i][0]) for i in range(5)): continue
            used = {r0, *cols}
            for rows in permutations(r for r in d if r not in used):
                rows = (r0,) + rows
                if not all(valid(rows[i][j], cols[j][i]) for i in range(5) for j in range(5)): continue
                print(rows, cols)
```
