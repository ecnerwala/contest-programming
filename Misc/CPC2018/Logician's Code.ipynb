```python
mag = ['Blaine', 'Copperfield', 'Houdini', 'Penn', 'Teller']
ass = ['Blackstone', 'Bundy', 'McGee', 'Pendragon', 'Smith']
pro = ['cards', 'hat', 'rabbit', 'scarf', 'wand']
```

```python
from itertools import *
from functools import *
for ast in permutations(ass):
    for prp in permutations(pro):
        try:
            assert len(ast[mag.index('Teller')]) > 5
            assert prp[mag.index('Penn')] in ['hat', 'scarf']
            assert prp[ast.index('Blackstone')] == 'wand'
            assert ast[mag.index('Penn')] == 'Pendragon'
            assert len(prp[ast.index('Smith')]) == 5
            assert len(prp[ast.index('McGee')]) == 5
            assert len(set('Blaine'.lower()) & set(ast[mag.index('Blaine')].lower()) & set(prp[mag.index('Blaine')])) == 1
            assert mag[ast.index('Smith')] in ['Blaine', 'Copperfield']
            assert prp[mag.index('Copperfield')] in ['hat', 'cards']
            print(list(zip(mag, ast, prp)))
        except AssertionError:
            continue
```
