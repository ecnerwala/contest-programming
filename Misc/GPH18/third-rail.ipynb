```python
with open("/home/andrew/Dropbox/dictionaries/sowpods.txt") as f:
    words = list(map(str.lower, map(str.strip, f.readlines())))
```

```python
val = {
    'e': 1,
    'a': 1,
    'i': 1,
    'o': 1,
    'n': 1,
    'r': 1,
    't': 1,
    'l': 1,
    's': 1,
    'u': 1,
    'd': 2,
    'g': 2,
    'b': 3,
    'c': 3,
    'm': 3,
    'p': 3,
    'f': 4,
    'h': 4,
    'w': 4,
    'v': 4,
    'y': 4,
    'k': 5,
    'j': 8,
    'x': 8,
    'q': 10,
    'z': 10,
}
assert len(val) == 26
```

```python
def score(word):
    return sum(map(val.get, word))
```

```python
from collections import Counter
def repeated(word):
    return Counter(word).most_common(1)[0][1]
```

```python
def get_word(nlen, nrep, nscore):
    return [w for w in words if len(w) == nlen and score(w) == nscore and repeated(w) == nrep]
```
