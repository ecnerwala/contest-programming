```python
from pymouse import PyMouse
import time
```

```python
m = PyMouse()
x_dim, y_dim = m.screen_size()
```

```python
def go(x, y, times):
    assert len(times) % 2 == 0
    for wait, hold in zip(times[0::2], times[1::2]):
        time.sleep(wait)
        m.press(x, y, 1)
        time.sleep(hold)
        m.release(x, y, 1)
```

```python
go(x_dim//2, y_dim//2, [3, 3, 3, 3, 3, 3, 3, 3])
```
