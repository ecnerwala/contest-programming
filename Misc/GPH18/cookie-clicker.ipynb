```python
from pymouse import PyMouse
import time
```

```python
m = PyMouse()
x_dim, y_dim = m.screen_size()
```

```python
def go(x, y, scale, times):
    for wait, hold in times:
        time.sleep(scale*wait)
        m.press(x, y, 1)
        time.sleep(scale*hold)
        m.release(x, y, 1)
```

```python
go(x_dim//2, y_dim//2, 1, [(1,1)] * 16)
```
