```python
with open('p-a.txt') as f:
    boxes = f.readlines()
    boxes = [i.strip().split() for i in boxes]
    boxes = [i[:4] + [i[6]] + [int(i[8][:-1])] for i in boxes]
```

```python
from collections import Counter
```

```python
levels = Counter(a[3] for a in boxes)
levels = {k: v / len(boxes) for k, v in levels.items()}
len(levels), levels
```
