```python
phone_digits = list(map(str, [9376, 663, 896, 84733, 3687, 3483, 749, 73836, 34448, 6463]))
pref_dig = {d[:3]: i for i, d in enumerate(phone_digits)}
assert len(pref_dig) == len(phone_digits) == 10
def phone_spell(v):
    return int(''.join(phone_digits[int(d)] for d in str(v)))
def phone_unspell(v):
    orig = v
    v = str(v)
    r = 0
    while v:
        if len(v) < 3: return orig
        d = pref_dig.get(v[:3], None)
        if d is None: return orig
        s = phone_digits[d]
        if not v.startswith(s): return orig
        v = v[len(s):]
        r = r*10+d
    return r
assert phone_unspell(6639376749) == 106
assert phone_unspell(1234) == 1234
```

```python
def do_ops(V):
    def is_good(V):
        return max(V) <= 10 ** 30
    def helper():
        # Increment
        for i in range(4):
            r = list(V)
            r[i] += 1
            yield tuple(r)

        # Type a
        for i in range(4):
            for j in range(4):
                r = list(V)
                r[i] += r[j]
                yield tuple(r)

        # Type b
        d, m = max((v, i) for i, v in enumerate(V))
        d //= 2
        for i in range(4):
            r = list(V)
            r[m] -= d
            r[i] += d
            yield tuple(r)

        # Type c
        yield tuple(phone_unspell(v) for v in V)
    yield from filter(is_good, map(tuple, map(sorted, helper())))
```

```python
START = (314159265, 358979323, 846264338, 327950288)
```

```python
import heapq
vis = {START}
q = [(-1, 0, START)]
while q:
    print(q[0])
    _, d, cur = heapq.heappop(q)
    for nxt in do_ops(cur):
        if nxt in vis: continue
        vis.add(nxt)
        heapq.heappush(q, (sum(nxt)*1e-8 + d+1, d+1, nxt))
```

```python
def print_path(V, t, i1):
    V = list(V)
    t = list(t)
    assert(all(V[i] <= t[i] for i in range(4)))
    assert t[i1] % V[i1] == 0
    assert (t[i1] // V[i1]) & (t[i1] // V[i1] - 1) == 0
    print(f"Start at: {V}")
    while V[i1] < t[i1]:
        for j in [0,1,2,3]:
            if j == i1: continue
            if (t[j] - V[j]) & V[i1]:
                V[j] += V[i1]
                print(f"A {i1+1}th into {j+1}th: {V}")
        V[i1] *= 2
        print(f"A double the {i1+1}th: {V}")
    while sum(V) < sum(t):
        for j in [0,1,2,3]:
            if j == i1: continue
            if t[j] > V[j]:
                V[j] += V[i1]
                print(f"A {i1+1}th into {j+1}th: {V}")
    print(f"C into: {[phone_unspell(i) for i in V]}")
print_path([370532258, 369230991, 1, 369223860], [646384733, 646384733, 268435456, 646384733], 2)
```
