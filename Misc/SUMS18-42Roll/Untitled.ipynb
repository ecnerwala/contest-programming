```python
dirs = {'L': (0, -1), 'R': (0, 1), 'U': (-1, 0), 'D': (1, 0)}
opp = {'U': 'D', 'D': 'U', 'L': 'R', 'R': 'L'}
def make_move(state, num):
    die, pos = state
    assert num != die['']
    loc = next(c for c in die if die[c] == num)
    npos = (pos[0] + dirs[loc][0], pos[1] + dirs[loc][1])
    if loc in 'UD':
        res = {'L': die['L'], 'R': die['R']}
        res[''] = die[loc]
        res[opp[loc]] = die['']
        return res, npos
    elif loc in 'LR':
        res = {}
        res[''] = die[loc]
        res[opp[loc]] = die['']
        if 'U' in die:
            res[loc] = die['U']
            res['D'] = die[opp[loc]]
        elif 'D' in die:
            res[loc] = die['D']
            res['U'] = die[opp[loc]]
        return res, npos
    else: assert False
def solve(s):
    s = list(map(int, s))
    a = s[0]-1
    state = ({'': a%4+1, 'D': (a+1)%4+1, 'L': (a+2)%4+1, 'R': (a+3)%4+1}, (0,0))
    res = {state[1]}
    for c in s[1:]:
        state = make_move(state, c)
        res.add(state[1])
    min_r = min(r for r,c in res)
    max_r = max(r for r,c in res)
    min_c = min(c for r,c in res)
    max_c = max(c for r,c in res)
    print('\n'.join(''.join('S' if (r,c) == (0,0) else ' X'[(r,c) in res] for c in range(min_c, max_c+1)) for r in range(min_r, max_r+1)))
    print()
    #return res
    
solve('1324314312312413241324132142341234123')
solve('43214321431234213421342134231431421342')
solve('34213421342134123412341241234123412314231423142')
solve('324132413243124134123412341231432413241324132')
solve('1324314312312413241324132142341234123')
solve('4234123412342142134213421342132142')
solve('1231421341321432143')
solve('2134213421342342134213421342134231423413241324')
```
