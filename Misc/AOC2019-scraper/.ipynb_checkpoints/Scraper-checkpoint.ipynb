```python
import requests
from bs4 import BeautifulSoup
from functools import cache
import json
```

```python
@cache
def fetch_leaderboard(day):
    resp = requests.get(f'https://adventofcode.com/2019/leaderboard/day/{day}')
    return resp.text
```

```python
fetch_leaderboard(1)
```

```python
@cache
def get_leaderboard_times(day):
    soup = BeautifulSoup(fetch_leaderboard(day), 'html.parser')
    entries = soup.find_all(class_='leaderboard-entry')
    res = []
    for i, e in enumerate(entries):
        pos = e.find(class_='leaderboard-position').string
        pos = pos.strip().removesuffix(')')
        pos = int(pos)
        assert pos == i % 100 + 1

        time = e.find(class_='leaderboard-time').string
        time = time.split('  ')[1]
        h,m,s = map(int, time.split(':'))
        res.append((h*60+m)*60+s)
    assert len(res) == 200
    res = [res[100:], res[:100]]
    return res
```

```python
print(get_leaderboard_times(1))
```

```python
@cache
def get_person_times(day, tgt):
    soup = BeautifulSoup(fetch_leaderboard(day), 'html.parser')
    entries = soup.find_all(class_='leaderboard-entry')
    res = [[], []]
    for i, e in enumerate(entries):
        elem = e.a
        name = next(elem.strings) if elem is not None and elem.string != '(AoC++)' else e.string
        
        if name != tgt: continue
        
        pos = e.find(class_='leaderboard-position').string
        pos = pos.strip().removesuffix(')')
        pos = int(pos)
        assert pos == i % 100 + 1

        time = e.find(class_='leaderboard-time').string
        time = time.split('  ')[1]
        h,m,s = map(int, time.split(':'))
        res[1 - (i//100)].append((pos, (h*60+m)*60+s))
    return res
```

```python
tot_leaderboard = 0
for d in range(1, 26):
    t = get_person_times(d, 'betaveros')
    print(d, t)
    if t[1]: tot_leaderboard += t[1][0][1]
tot_leaderboard += 701
print(tot_leaderboard // 3600)
```

```python
def get_world_record_segments():
    res = []
    for d in range(1, 26):
        s1, s2 = get_leaderboard_times(d)
        res.append(s1[0])
        res.append(s2[0] - s1[0])
    return res
```

```python
get_world_record_segments()
```

```python
def get_world_record_splits():
    cur = 0
    res = []
    for seg in get_world_record_segments():
        cur += seg
        res.append(cur)
    return res
```
