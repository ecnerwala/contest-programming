```python
def is_valid(x):
    try:
        assert len(x['byr']) == 4 and 1920 <= int(x['byr']) <= 2002
        assert len(x['iyr']) == 4 and 2010 <= int(x['iyr']) <= 2020
        assert len(x['eyr']) == 4 and 2020 <= int(x['eyr']) <= 2030
        hgt = x['hgt']
        if hgt.endswith('cm'):
            assert 150 <= int(hgt[:-2]) <= 193
        elif hgt.endswith('in'):
            assert 59 <= int(hgt[:-2]) <= 76
        else: assert False
        hcl = x['hcl']
        assert hcl[0] == '#' and len(hcl) == 7 and all(c in '0123456789abcdef' for c in hcl[1:])
        assert x['ecl'] in {'amb', 'blu', 'brn', 'gry', 'grn', 'hzl', 'oth'}
        assert len(x['pid']) == 9 and all(c in '0123456789' for c in x['pid'])
        return True
    except:
        return False
sum(map(is_valid, keys))
```
