```python
nums = [3,4,2,4,3,4,5,4]
seen = set()
def dfs(cur='A', path='', p=''):
    if (cur,path) in seen: return
    seen.add((cur,path))
    if len(path) == len(nums):
        yield path, p+cur
        return
    i = len(path)
    for nxt, w in adj[cur].items():
        #if nxt <= cur: continue
        yield from dfs(nxt, path+w[nums[i]-1], p+cur)
res = sum((list(dfs(i)) for i in adj), [])
print(len(res))
```
