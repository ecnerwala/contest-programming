```python
data = """2, 2, NORMAL, 759
2, 3, NORMAL, 180
2, 4, NORMAL, 821
2, 5, HEAVY, 977
2, 6, NORMAL, 331
2, 7, NORMAL, 701
2, 8, NORMAL, 859
2, 9, NORMAL, 508
2, 10, NORMAL, 616
2, 11, NORMAL, 503

5, 2, HEAVY, 583
5, 3, NORMAL, 222
5, 4, NORMAL, 244
5, 5, NORMAL, 509
5, 6, NORMAL, 179
5, 7, NORMAL, 100
5, 8, HEAVY, 998
5, 9, NORMAL, 190
5, 10, NORMAL, 756
5, 11, NORMAL, 879

8, 2, NORMAL, 336
8, 3, NORMAL, 594
8, 4, NORMAL, 51
8, 5, HEAVY, 727
8, 6, NORMAL, 219
8, 7, NORMAL, 49
8, 8, NORMAL, 552
8, 9, NORMAL, 905
8, 10, HEAVY, 927
8, 11, NORMAL, 212

11, 2, NORMAL, 237
11, 3, NORMAL, 55
11, 4, HEAVY, 385
11, 5, NORMAL, 740
11, 6, NORMAL, 987
11, 7, HEAVY, 908
11, 8, NORMAL, 964
11, 9, NORMAL, 71
11, 10, NORMAL, 539
11, 11, NORMAL, 828

14, 2, NORMAL, 294
14, 3, HEAVY, 649
14, 4, NORMAL, 841
14, 5, NORMAL, 246
14, 6, NORMAL, 54
14, 7, NORMAL, 938
14, 8, NORMAL, 212
14, 9, NORMAL, 989
14, 10, HEAVY, 973
14, 11, NORMAL, 12"""
data = [l.split(', ') for l in data.split('\n') if l]
grid = [
    [
        None for _ in range(12)
    ] * 5
    for _ in range(5)
]
for r,c,h,w in data:
    grid[int(r)//3][int(c)] = (int(w), h == 'HEAVY')
```

```python
res = 0
for a in range(12):
    for b in range(12):
        for c in range(12):
            for d in range(12):
                for e in range(12):
                    if 1 in (a,b,c,d,e): continue
                    t = 60 - 16 - 2*sum((a,b,c,d,e))
                    g= grid[0][:a+1] + grid[1][:b+1] + grid[2][:c+1] + grid[3][:d+1] + grid[4][:e+1]
                    g = [a for a in g if a]
                    #print(a,b,c,d,e,g,t)
                    if t <= 0: continue
                    nor = sorted(w for w,h in g if not h)[::-1]
                    hea = sorted(w for w,h in g if h)[::-1]
                    res = max(res, sum(nor[:t]))
                    for nh in range(len(hea)+1):
                        if 3 * nh <= t:
                            res = max(res, sum(hea[:nh]) + sum(nor[:t-3*nh]))
```
