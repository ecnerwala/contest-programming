```python
for a in range(1,16):
    for b in range(1,16):
        for c in range(1,16):
            for d in range(1,16):
                for e in range(1,16):
                    row = [a,b,c,d,e]
                    tot = []
                    while row:
                        tot.extend(row)
                        row = [abs(x-y) for x,y in zip(row, row[1:])]
                    if len(set(tot)) == len(tot) and tot[-3] == 9:
                        print(tot)
                        chrs = 'aeioulsntrbh9cm'
                        mp = {k:v for k, v in zip(tot, chrs)}
                        assert mp[9] == '9'
                        inds = [11,14,1,6,12]
                        print(''.join(mp[i] for i in inds))
```
