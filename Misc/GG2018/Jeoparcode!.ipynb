# Problem 2

```python
def p2():
    target = 10 ** 100
    fac = 1
    cur = 0
    while not fac > target:
        cur += 1
        fac *= cur
    return cur
p2()
```

```python
def factorial(i):
    return i * factorial(i-1) if i else 1
assert factorial(70) > 10 ** 100
assert not factorial(69) > 10 ** 100
```

# Problem 3

```python
def p3():
    num = 0x64
    def itoa(v, b):
        return itoa(v // b, b) + hex(v % b)[2] if v else ''
    assert itoa(num, 16) == '64'
    digits = ''.join(itoa(num, b) for b in range(2, 17))
    from collections import Counter
    cnt = Counter(digits)
    return ''.join(sorted(cnt, key=(lambda d: (cnt[d], d)),reverse=True))
p3()
```

# Problem 4

```python
from itertools import product
from functools import reduce
def p4():
    res = 0
    for a in product(*[[(0.9, 5), (0.1, 45)] for _ in range(10)]):
        #print(a)
        time = sum(t for p, t in a)
        if time <= 90:
            prob = reduce((lambda x,y:x*y), (p for p,t in a))
            res += prob
    r2 = ((0.9) ** 10 + (0.9) ** 9 * 0.1 * 10)
    assert abs(res-r2) < 1e-9
    return "{0:.5f}".format(res)
p4()
```

# Problem 5

```python
def p5():
    w = [ord(c)-ord('a') for c in 'ekrktsir']
    for i in range(128):
        w[i % len(w)] += i+1
    return ''.join(chr(ord('a') + i % 26) for i in w)
p5()
```
