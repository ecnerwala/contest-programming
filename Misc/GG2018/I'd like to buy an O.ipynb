```python
passage = """’Twas brillig, and the slithy toves
Did gyre and gimble in the wade;
All mimsy were the borogoves,
And the mome raths outgrabe.
"Beware the Jabberwock, my son!
The jaws that bite, the claws that catch!
Beware the Jubjub bird, and shun
The frumious Bandersnatch!"""

from string import ascii_lowercase as alpha
vowels = 'aeiou'
cons = ''.join(c for c in alpha if c not in vowels)

from collections import defaultdict
scores = defaultdict(int)
for v in vowels:
    scores[v] = 1
for i,c in enumerate(cons):
    scores[c] = i+1
for c in alpha:
    scores[c.upper()] = 2 * scores[c]
print(scores)

assert sum(scores[c] for c in 'Cat') == 21
assert sum(scores[c] for c in 'cat') == 19
assert sum(scores[c] for c in 'The Cat in the Hat') == 124
print(sum(scores[c] for c in passage))
```
