```python
def solve(inp):
    g = [[int(v) for v in l.split(' ')] for l in inp.split('\n')]
    dp = [0] * len(g[0])
    for r in g:
        dp = [max(dp[max(0,i-1):min(len(dp),i+2)]) for i in range(len(dp))]
        for i,v in enumerate(r):
            dp[i] += v
    return max(dp)
```

```python
def solvejump(inp):
    rows = inp.split('\n')
    return max(solve('\n'.join(rows[:s])) + solve('\n'.join(rows[s:])) for s in range(1, len(rows)))
```
