```python
W = 9
def get_probs(s):
    a = [0.] * W
    a[s] = 1.
    for r in range(6):
        a = [i/2 for i in a]
        b = list(map(sum, zip(a[:-1], a[1:])))
        b[0] += a[0]
        b[-1] += a[-1]
        b = [i/2 for i in b]
        a = [b[0]] + list(map(sum, zip(b[:-1], b[1:]))) + [b[-1]]
    return a
get_probs(0)
```

```python
def argmax(s):
    return max(range(len(s)), key=lambda i:s[i])
```

```python
print(get_probs(0))
```

```python
scores = [100, 500, 1000, 0, 10 * 1000, 10, 1000, 500, 100]

best = {}

def dp(cur=()):
    if cur in best:
        return best[cur]
    if len(cur) == 4:
        best[cur] = sum(scores[i] for i in set(cur)), -1
    else:
        v, ind = -1, -1
        for i in range(9):
            val = sum(dp(cur+(b,))[0] * p for b,p in enumerate(get_probs(i)))
            if val > v:
                v, ind = val, i
        best[cur] = v, ind
    return best[cur]
dp()
```
