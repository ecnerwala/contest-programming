```python
opts = """MOVE 1
MOVE 1
MOVE 2
MOVE 2
MOVE 3
BACK UP 1
TURN LEFT
TURN RIGHT
U-TURN
TELEPORT!""".split('\n')

from itertools import *
from collections import *

# Up, Right, Down, Left
deltas = [(-1, 0), (0, 1), (1, 0), (0, -1)]

def move_wall(grid, st, d):
    d %= 4
    x,y = st
    ex,ey = x + deltas[d][0], y+deltas[d][1]
    wall = tuple(sorted([(x,y), (ex,ey)]))
    if wall in grid['walls']:
        return x,y
    return ex,ey

def do_move(grid, loc, cmd):
    (x,y), d = loc
    if cmd.startswith("MOVE"):
        dist = int(cmd.split()[-1])
        for _ in range(dist):
            x,y = move_wall(grid,(x,y), d)
    elif cmd == "BACK UP 1":
        x,y = move_wall(grid, (x,y), (d + 2)%4)
    elif cmd == 'TURN LEFT':
        d -= 1
    elif cmd == 'TURN RIGHT':
        d += 1
    elif cmd == 'U-TURN':
        d += 2
    elif cmd == 'TELEPORT!':
        x += deltas[d][0] * grid['tp']
        y += deltas[d][1] * grid['tp']
    d %= 4
    if grid['squares'][x][y] == 'L':
        d -= 1
    elif grid['squares'][x][y] == 'R':
        d += 1
    return (x,y), d

def run(grid, cmds):
    (x,y) = next((x,y) for x in range(5) for y in range(5) if '0' <= grid['squares'][x][y] <= '3')
    d = int(grid['squares'][x][y])
    print((x,y),d)
    res = []
    for cmd in cmds:
        (x,y),d = do_move(grid, ((x,y), d), cmd)
        if not (0 <= x < 5 and 0 <= y < 5):
            return None
        res.append((x, y, grid['letters'][x][y]))
    if grid['squares'][x][y] != 'W':
        return None
    return res

def make_grid(squares, horiz, vert, letters, tp):
    return {
        'squares': squares.split('\n'),
        'walls': {((x-1,y), (x,y)) for x,y in horiz} + {((x,y-1), (x,y)) for x,y in vert},
        'letters': letters.split('\n'),
        'tp': tp,
    }

def solve(g1, g2):
    for cmds in permutations(opts, 5):
        r1, r2 = run(g1, cmds), run(g2, cmd2)
        if r1 is None or r2 is None:
            continue
        print(r1, r2)
```
